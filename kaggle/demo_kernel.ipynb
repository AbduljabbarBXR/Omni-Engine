```python
import os, subprocess, sys, glob
for root, dirs, files in os.walk('/kaggle/input'):
    print('DIR', root, '|', files[:10])
run_dir = glob.glob('/kaggle/input/**/cfg.json', recursive=True)[0].rsplit('/', 1)[0]
print('run dir:', run_dir)
prompts = [
    "To be, or not to be",
    "All the world's a stage",
    "The course of true love never did run smooth",
    "What light through yonder window breaks",
]
for p in prompts:
    cmd = [sys.executable, '-m', 'omni.generate',
           '--run', run_dir,
           '--base', 'EleutherAI/pythia-160m-deduped',
           '--prompt', p, '--tokens', '60', '--temperature', '0.8']
    r = subprocess.run(cmd, capture_output=True, text=True)
    print(r.stdout[-1200:])
    if r.returncode != 0:
        print(r.stderr[-800:])
print('DEMO_DONE')
```
